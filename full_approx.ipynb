```python
from Monte_Carlo_sim import run_mc
from plotting import animate_run
import numpy as np

# --- Parameters (edit these) ---
target_term = "second"   # "first", "second", or "full"

L = 200
N_plus = 20
N_minus = 20

a = 1.0
xi = 1.0
xiF = 5.0

T = 1.0
kappa = 1.0      # K / T
e0_red = 5.0     # E0 / T

K = kappa * T
E0 = e0_red * T

step_size = 0.6
n_sweeps = 5000
sample_every = 50
seed = 0

# --- Run ---
charges = np.array([+1] * N_plus + [-1] * N_minus, dtype=int)
N = len(charges)

print(f"Running simulation for target_term = {target_term!r} ...")
snapshots, E_trace, pair_trace, acc_rate = run_mc(
    L=L,
    N=N,
    charges=charges,
    K=K,
    T=T,
    E0=E0,
    a=a,
    xi=xi,
    xiF=xiF,
    step_size=step_size,
    n_sweeps=n_sweeps,
    sample_every=sample_every,
    target_term=target_term,
    seed=seed,
)
print(f"Done. Acceptance rate = {acc_rate:.3f}")

if target_term == "first":
    display_value = kappa
    display_label = "K/T"
elif target_term == "second":
    display_value = e0_red
    display_label = "E0/T"
else:
    display_value = kappa
    display_label = "K/T"

animate_run(
    snapshots, E_trace, pair_trace, charges, L,
    display_value, display_label, sample_every
)
```
