```python
import numpy as np
import matplotlib.pyplot as plt
from Monte_Carlo import run_mc1
from Visualization import animate_snapshots1, plot_final_snapshot1

seed = 0
rng = np.random.default_rng(seed)

N = 40
assert N % 2 == 0

q = np.ones(N, dtype=int)
q[N // 2:] = -1

rng = np.random.default_rng(seed)
rng.shuffle(q)

print("sum(q) =", np.sum(q))
print("N_plus =", np.sum(q == 1))
print("N_minus =", np.sum(q == -1))

a = 0.02
step_size = 0.04
kappa = 1

alpha = 5
n_cut = 1
m_cut = 10

n_burnin = 500
n_sweeps = 10000
sample_every = 100
```

```python
snapshots, U1_trace, acceptance_rate = run_mc1(
    N=N,
    q=q,
    kappa=kappa,
    a=a,
    alpha=alpha,
    n_cut=n_cut,
    m_cut=m_cut,
    step_size=step_size,
    n_sweeps=n_sweeps,
    sample_every=sample_every,
    n_burnin=n_burnin,
    seed=seed,
)

print("acceptance_rate =", acceptance_rate)

animate_snapshots1(
    snapshots=snapshots,
    q=q,
    sample_every=sample_every,
    kappa=kappa,
)
```

```python
plot_final_snapshot1(
    snapshots=snapshots,
    q=q,
    sample_every=sample_every,
    kappa=kappa,
)
```
